```python
# Jawaban Latihan 4

#tugas A

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) cetak pembacaan pertama dan terakhir memakai indeks
print(f"Pembacaan pertama: {tegangan_harian[0]}")
print(f"Pembacaan terakhir: {tegangan_harian[-1]}")

# (b) tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)
print(f"List setelah ditambahkan: {tegangan_harian}")

# (c) cetak jumlah total data dengan len()
print(f"Jumlah total data: {len(tegangan_harian)}")
```

```python
# Jawaban Latihan 5

# Buat dictionary spesifikasi motor listrik
motor_listrik = {
    'nama': 'Motor AC Induksi',
    'daya_kW': 7.5,
    'tegangan_V': 380,
    'rpm': 1450
}

# Cetak salah satu nilai lewat key-nya (misal: daya_kW)
print(f"Daya motor: {motor_listrik['daya_kW']} kW")

# Ubah nilai rpm
motor_listrik['rpm'] = 1500

# Cetak ulang seluruh dictionary setelah perubahan
print("Spesifikasi motor setelah diupdate:")
print(motor_listrik)
```

```python
# Jawaban Latihan 6
pembangkit = [
    {"PLTU": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"PLTS": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"PLTA": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# (a) Cetak nama unit kedua
print(f"PLTS: {pembangkit[1]['PLTS']}")

# (b) Ubah status salah satu unit (misal: Unit-3 menjadi 'siaga')
pembangkit[2]['status'] = 'siaga'
print(f"Status Unit-3 setelah diubah: {pembangkit[2]['status']}")

# (c) Tambahkan satu unit baru ke list dengan append()
unit_baru = {"PLTB": "Unit-4", "kapasitas_MW": 120, "status": "uji coba"}
pembangkit.append(unit_baru)

# (d) Cetak seluruh data pembangkit
print("\nSeluruh data pembangkit setelah diupdate:")
for unit in pembangkit:
    print(unit)
```
